```python
i = 0
result1 = 0
result = 0  # Initialize result to 0 before the loop
while i < 100:
  i = i + 1
  if i % 2 == 0:
    result1 = result + i

print('1번 방법 : {0}'.format(result1))
```

```python
# 2번 방법 (break 이용)
j = 0
result2 = 0
while True:
  if j > 100:
    break

    j = j + 1
    if j % 2 == 0:
      result2 = result2 + j

print('2번 방법 (break) : {0}'.format(result2))
```
