```python
print("화면에 직접 출력")
print('ab\'c')
print("ab'c")
print("deosen't")
print('does')
print('doesn\'t')
print("'string'")
print("\"string\"")
print('"string"')
```

```python
s1 = '화면에 직접 출력'
s2 = 'ab\'c'
s3 = 'does'
print(s1)
print(s2)
print(s3)
print(s1[0]) # 첫 글자
print(s2[1]) # 두 번째 글자
print(s3[1:3])
print(s3[-1]) # 마지막 글자
print(s3[-2]) # 뒤에서 두 번째
```

```python
item1 = '사과'
price1 = 1000
item2 = '바나나'
price2 = 500
print(item1, price1)
print(item2, price2)
```

```python
item1 = '사과'
price1 = 1000
item2 = '바나나'
price2 = 500
str1 = '{0}은 {1}원입니다.'
print(str1.format(item1, price1))
print(str1.format(item2, price2))
```

```python
print(item1, price1, sep='*', end='/')
print(item2, price2)
```

```python
str2 = '%s는 %d원입니다.'
print(str2%(item1, price1))
print(str2%(item2, price2))
```
