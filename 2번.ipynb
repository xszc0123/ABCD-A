```python
results = model('yolo2.png')  # 업로드된 파일명 정확히 사용

results[0].show()

# 고양이 클래스 ID: 15
cat_class_id = 15
cat_count = sum(1 for box in results[0].boxes.cls if box.item() == cat_class_id)

print(f"사진 속 고양이 수: {cat_count}마리")
```
