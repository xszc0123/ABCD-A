```python
#압축 해제
import zipfile

with zipfile.ZipFile('yolov8.v1i.yolov8.zip', 'r') as zip_ref:
    zip_ref.extractall('dataset')
```

```python
import os

# valid/images 및 valid/labels 폴더 생성
os.makedirs('dataset/valid/images', exist_ok=True)
os.makedirs('dataset/valid/labels', exist_ok=True)
```

```python
import os
import shutil

src_images = 'dataset/train/images'
src_labels = 'dataset/train/labels'
dst_images = 'dataset/valid/images'
dst_labels = 'dataset/valid/labels'

os.makedirs(dst_images, exist_ok=True)
os.makedirs(dst_labels, exist_ok=True)

# 이미지 전체 복사
for fname in os.listdir(src_images):
    if fname.endswith(('.jpg', '.png', '.jpeg')):
        shutil.copy(os.path.join(src_images, fname), os.path.join(dst_images, fname))
        label = os.path.splitext(fname)[0] + '.txt'
        shutil.copy(os.path.join(src_labels, label), os.path.join(dst_labels, label))

print("✅ 모든 이미지와 라벨이 valid로 복사되었습니다.")
```

```python
import cv2
import os
from glob import glob

img_folder = 'runs\segment\predict'
output_video = 'segmentation_result2.mp4'

# 이미지 목록
img_files = sorted(glob(os.path.join(img_folder, '*.jpg')) + glob(os.path.join(img_folder, '*.png')))
if not img_files:
    print("❌ 결과 이미지가 없습니다.")
    exit()

# 비디오 설정
frame = cv2.imread(img_files[0])
height, width, _ = frame.shape
video = cv2.VideoWriter(output_video, cv2.VideoWriter_fourcc(*'mp4v'), 5, (width, height))

for img in img_files:
    video.write(cv2.imread(img))

video.release()
print(f"✅ 비디오 저장 완료: {output_video}")
```

```python
from ultralytics import YOLO
import cv2
import os

model_path = 'C:/Users/user/Desktop/obstacle/runs/segment/train/weights/best.pt'
video_path = 'C:/Users/user/Desktop/obstacle/11.mp4'  # 경로 오타 수정

# 경로 확인
if not os.path.exists(model_path):
    print("모델 경로가 잘못됨")
    exit()

if not os.path.exists(video_path):
    print("비디오 경로가 잘못됨")
    exit()

model = YOLO(model_path, task='segment')
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("비디오를 열 수 없음")
    exit()

# 비디오 저장 설정
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('output_video.mp4', fourcc, 30.0, (int(cap.get(3)), int(cap.get(4))))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print("프레임을 더 이상 읽을 수 없음")
        break

    results = model(frame)
    annotated_frame = results[0].plot()

    # 확인용 출력
    print(f"Frame processed. Boxes: {results[0].boxes}")

    cv2.imshow('Segmented Frame', annotated_frame)
    out.write(annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```
